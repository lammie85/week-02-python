```python
lucky_number = 15
attempts_left = 3

while attempts_left > 0:
    guess = int(input("Please Guess a number: "))

    if guess == lucky_number:
        print("Congratulations you are correct!")
        break

    attempts_left = attempts_left - 1

    if guess > lucky_number:
        print("It's lower!")
    elif guess < lucky_number:
        print("It's higher!")

    if guess > 10 and guess < 15 and guess != 15:
        print("You are almost there!")

    if attempts_left == 0:
        print("Game Over!")
```

```python
#FUNCTIONS: they are like verbs in coding

#SImple interest : Principal * rate * time / 100

principal = 4000000
rate = 16
time = 3

simple_interest = (principal * rate * time) / 100

print(simple_interest)
```

```python
def simple_interest (principal, rate, time):
    simp = (principal * rate * time)/ 100
    return simp
```

```python
#calling a function
January_SI = simple_interest(500000, 18, 6)
February = simple_interest(1000000, 21, 10)

print(January_SI)
print("YOur interest after 10 years is: ", February)
```
