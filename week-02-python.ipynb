```python
import json

customer = {
    "name": "Emmanuel Lammy",
    "email": "emmanuel@yahoo.com",
    "phone": "08123456789",
    "items": ["laptop", "mouse", "keyboard"]
}
#save
with open("customers.json", 'w') as f:
    json.dump(customer, f, indent=2)

#load
with open("customers.json") as f:
    json_data = json.load(f)
    print(json_data)
```
